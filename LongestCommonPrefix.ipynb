```python
strs = ["flower","flow","flight","flght"]
class Solution(object):
    def longestCommonPrefix(self, strs):
        i = 0
        j = 0
        cmn = ""
          
        if len(min(strs, key=len)) == 0 :
            return ""
        if len(strs) == 1:
            return strs[0]
        
        while i < len(strs) and j <= len(min(strs, key=len)) -1:
            if i != len(strs) -1:
                if strs[i][j] == strs[i+1][j]:
                    i +=1
                else:
                    return cmn
            else:
                cmn += strs[i][j]
                i=0
                j+=1
        return cmn

Solution.longestCommonPrefix(None, strs)
```
