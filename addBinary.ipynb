```python
class Solution(object):
    def addBinary(self, a, b):
        """
        :type a: str
        :type b: str
        :rtype: str
        """
        i = 1
        a_int = 0
        for j in a[::-1]:
            old = i
            if j == "1":
                a_int += i
            i = old * 2
        
        i = 1     
        b_int = 0
        for j in b[::-1]:
            old = i
            if j == "1":
                b_int += i
            i = old * 2
        return bin(a_int + b_int).replace("0b", "")
        
```
